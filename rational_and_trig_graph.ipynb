```python
%%manim -v WARNING -ql  -i rational_Curve_Sketching
class rational_Curve_Sketching(MovingCameraScene):
    def construct(self):

        # Axes for graph 
        axes = Axes(
            x_range=[-5, 5, 1],
            y_range=[-5, 5, 1],
            x_length=10,
            y_length=10,
            axis_config={"color": WHITE},
            tips=False
        )
        
        axes.scale(0.8)  # Slightly smaller so it fits nicely
        axes.move_to(ORIGIN)

##############################################################################################################

        
        # Plot key points when graph cuts y-axis
        y_axis_intercept = Dot(axes.c2p(0, 1/2), color=RED)
        y_axis_label = MathTex("(0, \\frac{1}{2})").next_to(y_axis_intercept, RIGHT + UP * 0.1).scale(0.7)
        self.wait(2)

        # Asymptote line y=0


        asymptote = DashedLine(start=axes.c2p(-5, 0), end=axes.c2p(5, 0), color=YELLOW)
        asymptote_label = MathTex("y = 0", color=YELLOW).next_to(asymptote, DOWN).scale(0.9).move_to(axes.c2p(-2, 3))
        text_asymptote =  MathTex("Asymptote", color=YELLOW).next_to(asymptote_label, DOWN)



        #Asymptoe x = -2
        asymptote1 = DashedLine(start=axes.c2p(-2, -5), end=axes.c2p(-2, 5), color=YELLOW)
        asymptote1_label = MathTex("x = -2", color=YELLOW).next_to(asymptote1, LEFT).scale(0.9)
        text_asymptote1 =  MathTex("Asymptote", color=YELLOW).next_to(asymptote1_label, DOWN)   


        # Add arrow pointing to x tends to infinity region and y tends to 0 
        
        arrow1 = Arrow(
            start=axes.c2p(1, 0),
            end=axes.c2p(5, 0),
            color=GREEN,
            buff=0.1,
           
        )  
       
        label11 = MathTex("x", r"\to", r"\infty^+", color=ORANGE).next_to(arrow1, DOWN)

        arrow11 = Arrow(
            start=axes.c2p(5, 5),
            end=axes.c2p(5, 0.1),
            color=ORANGE,
            buff=0.1,
        )  
        label111 = MathTex("y", r"\to", r"0^+", color=ORANGE).next_to(arrow11, RIGHT)   
        # Add arrow pointing to x tends to -2 from right side

        arrow2 = Arrow(
            start=axes.c2p(1, 0),
            end=axes.c2p(-1.9, 0),
            color=ORANGE,
            buff=0.1,
        )  
        label12 = MathTex("x", r"\to", r"-2^+", color =ORANGE).next_to(arrow2, DOWN)

        arrow22 = Arrow(
            start=axes.c2p(-1, 0.5),
            end=axes.c2p(-1, 4),  
            color=ORANGE,
            buff=0.1,
        )  
        label122 = MathTex("y", r"\to", r"\infty", color =ORANGE).next_to(arrow22, LEFT)    

        

        # Adding arrow pointing to x tends to -infinity from left side
        arrow3 = Arrow(
            start=axes.c2p(-3, 0),
            end=axes.c2p(-5, 0),
            color=ORANGE,
            buff=0.1,
        )  
        label13 = MathTex("x", r"\to", r"-\infty", color =ORANGE).next_to(arrow3, DOWN)



        # Graph of the function y = 1/(x+2) from x = 0 to x = 5
        graph1 = axes.plot(lambda x: 1/(x + 2), x_range=[0, 5], color=BLUE)
        graph1_label = MathTex(r"y = \frac{1}{x+2}", color=BLUE).next_to(graph1, UP * 0.5 + RIGHT).scale(0.7)   

        # Graph of the function y = 1/(x+2) from x = 0 to x = -1.9
        graph2 = axes.plot(lambda x: 1/(x + 2), x_range=[-1.9, 0], color=BLUE)
    








    #####################################################################################################################################

        # Sequence of animations 

        self.play(Create(axes), run_time=2)
        self.wait(1)
        self.play(Create(y_axis_intercept), Write(y_axis_label), run_time=2)
        self.wait(2)

        # Draw asymptote y=0
        self.play(Create(asymptote), run_time=2)
        self.play(Write(asymptote_label), Write(text_asymptote), run_time=1)
        self.wait(2) 
        self.play(FadeOut(asymptote_label), FadeOut(text_asymptote))
        self.wait(1)  


        # Shake animation for asymptote y=0
        for _ in range(3):
            self.play(asymptote.animate.shift(UP * 0.05), run_time=0.05)
            self.play(asymptote.animate.shift(DOWN * 0.05), run_time=0.05)
        self.wait(2)

        # Draw asymptote x = -2
        self.play(Create(asymptote1), run_time=2)
        self.play(Write(asymptote1_label), Write(text_asymptote1), run_time=1)
        self.wait(2) 
        self.play(FadeOut(asymptote1_label), FadeOut(text_asymptote1))
        self.wait(1)


        # Shake animation for asymptote x = -2
        for _ in range(3):
            self.play(asymptote1.animate.shift(RIGHT * 0.05), run_time=0.05)
            self.play(asymptote1.animate.shift(LEFT * 0.05), run_time=0.05)
        self.wait(2)

        # Add arrow and label for x>0 region
        self.play(Create(arrow1), Write(label11), run_time=2)
        self.play(Create(arrow11), Write(label111), run_time=2)
        self.wait(3)
        
        
        

        # Draw graph of the function y = 1/(x+2) from x = 0 to x = 5
        self.play(Create(graph1), run_time=2)
        self.wait(2)
        self.play(Write(graph1_label), run_time=1)
        self.wait(2)
        self.play(FadeOut(arrow1), FadeOut(label11), FadeOut(arrow11), FadeOut(label111), run_time=2)


        # Add arrow and label for x tends to -2 from right side
        self.play(Create(arrow2), Write(label12), run_time=2)
        self.wait(3)
        self.play(Create(arrow22), Write(label122), run_time=2)
        self.wait(3)
        self.play(FadeOut(arrow2), FadeOut(label12), run_time=2)
        self.play(FadeOut(arrow22), FadeOut(label122), run_time=2)

        # Draw graph of the function y = 1/(x+2) from x = 0 to x = -1.9
        self.play(Create(graph2), run_time=2)
        self.wait(2)
        self.play(Write(graph1_label), run_time=1)
        self.wait(2)
        





```

```python
%%manim -v WARNING -ql  -i WeierstrassVisualization


class WeierstrassVisualization(Scene):
    def construct(self):
        # Title
        title = MathTex(
            r"f(x) = \sum_{n=0}^{\infty} a^n \cos(b^n \pi x)",
            color=YELLOW
        ).to_edge(UP)
        self.play(Write(title))
        self.wait(1)

        # Axes
        axes = Axes(
            x_range=[-2, 2, 1],
            y_range=[-2, 2, 1],
            x_length=8,
            y_length=4,
            tips=False,
        ).to_edge(DOWN)
        labels = axes.get_axis_labels(x_label="x", y_label="f(x)")
        self.play(Create(axes), Write(labels))
        self.wait(1)

        # Parameters for the Weierstrass function
        a = 0.5
        b = 3
        n_terms = 50

        def weierstrass(x):
            return sum(a**n * np.cos((b**n) * np.pi * x) for n in range(n_terms))

        # Generate points
        x_vals = np.linspace(-2, 2, 2000)
        y_vals = [weierstrass(x) for x in x_vals]
        graph = axes.plot_line_graph(x_vals, y_vals, line_color=BLUE)

        # Animate the graph drawing
        self.play(Create(graph, run_time=5, rate_func=linear))
        self.wait(1)

        # Highlight its roughness
        rough_text = Text(
            "Continuous everywhere, Differentiable nowhere!",
            font_size=28,
            color=ORANGE
        ).next_to(title, DOWN)
        self.play(Write(rough_text))
        self.wait(2)
```

```python
%%manim -v WARNING -ql  -i DynamicWeierstrass1


class DynamicWeierstrass1(MovingCameraScene):
    def construct(self):
        # --- Dark background ---
        self.camera.background_color = "#000000"

        # --- Title ---
        title = MathTex(
            r"f(x) = \sum_{n=0}^{50} a^n \cos(b^n \pi x)",
            color=YELLOW
        ).to_edge(UP)
        self.play(Write(title))

        # --- Axes setup ---
        axes = Axes(
            x_range=[-2, 2, 1],
            y_range=[-2, 2, 1],
            x_length=8,
            y_length=4,
            tips=False,
            axis_config={"color": GRAY}
        ).to_edge(DOWN)
        labels = axes.get_axis_labels(x_label="x", y_label="f(x)").set_color(GRAY)
        self.play(Create(axes), Write(labels))
        self.wait(0.5)

        # --- Parameters ---
        a = 0.5
        n_terms = 50
        x_vals = np.linspace(-2, 2, 1200)

        def weierstrass(x, b):
            return sum(a**n * np.cos((b**n) * np.pi * x) for n in range(n_terms))

        b_value = ValueTracker(2.0)

        def get_graph():
            y_vals = [weierstrass(x, b_value.get_value()) for x in x_vals]
            color = interpolate_color(BLUE, PURPLE, (b_value.get_value() - 2) / 5)
            return axes.plot_line_graph(x_vals, y_vals, line_color=color).set_stroke(width=1.8)

        graph = always_redraw(get_graph)

        # --- Smoothly fade in the curve instead of 'Create' ---
        self.add(graph)
        self.play(FadeIn(graph, shift=UP, run_time=2))
        self.wait(0.5)

        # --- Animate 'b' changing ---
        b_label = MathTex("b = 2.0", color=ORANGE).next_to(title, DOWN)
        self.play(Write(b_label))

        def update_label(obj):
            obj.become(
                MathTex(f"b = {b_value.get_value():.2f}", color=ORANGE).next_to(title, DOWN)
            )

        b_label.add_updater(update_label)

        self.play(b_value.animate.set_value(7.0), run_time=8, rate_func=smooth)
        self.wait(1)
        b_label.remove_updater(update_label)

        # --- Zoom in on the rough section ---
        zoom_area = Rectangle(
            width=2, height=1, color=YELLOW
        ).move_to(axes.c2p(0.5, 0))
        self.play(Create(zoom_area))
        self.wait(1)

        self.play(
            self.camera.frame.animate
            .scale(0.4)
            .move_to(axes.c2p(0.5, 0)),
            run_time=4,
            rate_func=smooth
        )

        self.wait(3)
```
