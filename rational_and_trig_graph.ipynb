```python
%%manim -v WARNING -ql -i rational_Curve_Sketching
class rational_Curve_Sketching(MovingCameraScene):
    def construct(self):

        # Axes for graph 
        axes = Axes(
            x_range=[-5, 5, 1],
            y_range=[-5, 5, 1],
            x_length=10,
            y_length=10,
            axis_config={"color": WHITE},
            tips=False
        )
        
        axes.scale(0.8)  # Slightly smaller so it fits nicely
        axes.move_to(ORIGIN)

##############################################################################################################

        
        # Plot key points when graph cuts y-axis
        y_axis_intercept = Dot(axes.c2p(0, 1/2), color=RED)
        y_axis_label = MathTex("(0, \\frac{1}{2})").next_to(y_axis_intercept, RIGHT + UP * 0.1).scale(0.7)
        

        extra_point = Dot(axes.c2p(-3, -1), color=RED)
        extra_point_label = MathTex("(-3, -1)").next_to(extra_point, LEFT + DOWN * 0.1).scale(0.7)
        

        # Asymptote line y=0


        asymptote = DashedLine(start=axes.c2p(-5, 0), end=axes.c2p(5, 0), color=YELLOW)
        asymptote_label = MathTex("y = 0", color=YELLOW).next_to(asymptote, DOWN).scale(0.8).move_to(axes.c2p(-2,1.5))
        text_asymptote =  MathTex("Asymptote", color=YELLOW).next_to(asymptote_label, DOWN)



        #Asymptoe x = -2
        asymptote1 = DashedLine(start=axes.c2p(-2, -5), end=axes.c2p(-2, 5), color=YELLOW)
        asymptote1_label = MathTex("x = -2", color=YELLOW).move_to(axes.c2p(-4,2)).scale(0.8)
        text_asymptote1 =  MathTex("Asymptote", color=YELLOW).next_to(asymptote1_label, DOWN)   


        # Add arrow pointing to x tends to infinity region and y tends to 0 
        
        arrow1 = Arrow(
            start=axes.c2p(1, 0),
            end=axes.c2p(5, 0),
            color=ORANGE,
            buff=0.1,
           
        )  
       
        label11 = MathTex("x", r"\to", r"\infty^+", color=ORANGE).next_to(arrow1, DOWN)

        arrow11 = Arrow(
            start=axes.c2p(5, 5),
            end=axes.c2p(5, 0.1),
            color=ORANGE,
            buff=0.1,
        )  
        label111 = MathTex("y", r"\to", r"0^+", color=ORANGE).next_to(arrow11, RIGHT)   
        # Add arrow pointing to x tends to -2 from right side and y tends to infinity

        arrow2 = Arrow(
            start=axes.c2p(1, 0),
            end=axes.c2p(-1.9, 0),
            color=ORANGE,
            buff=0.1,
        )  
        label12 = MathTex("x", r"\to", r"-2^+", color =ORANGE).move_to([-0.8, -0.5,1]).scale(0.8)

        arrow22 = Arrow(
            start=axes.c2p(-1.85, 0.5),
            end=axes.c2p(-1.85, 4),  
            color=ORANGE,
            buff=0.1,
        )  
        label122 = MathTex("y", r"\to", r"\infty", color =ORANGE).next_to(arrow22, LEFT)    

        

        # Adding arrow pointing to x tends to -infinity from left side
        arrow3 = Arrow(
            start=axes.c2p(-3, 0),
            end=axes.c2p(-5, 0),
            color=ORANGE,
            buff=0.1,
        )  
        label13 = MathTex("x", r"\to", r"-\infty", color =ORANGE).next_to(arrow3, DOWN)

        #Adding arrow pointing to x tends to -2 from left side and y tends to -infinity
        arrow4 = Arrow(
            start=axes.c2p(-5, 0),
            end=axes.c2p(-1.9, 0),  
            color=ORANGE,
            buff=0.1,
        )  
        label14 = MathTex("x", r"\to", r"-2", color =ORANGE).next_to(arrow4, UP)  

        arrow44 = Arrow(
            start=axes.c2p(-1, -0.5),
            end=axes.c2p(-1, -4),  
            color=ORANGE,
            buff=0.1,
        )  
        label144 = MathTex("y", r"\to", r"-\infty", color =ORANGE).next_to(arrow44, RIGHT)  

        #Adding arrow pointing to x tends to negative infinity  from left side and y tends to 0 from negative side
        arrow5 = Arrow(
            start=axes.c2p(-3, 0),
            end=axes.c2p(-5, 0),  
            color=ORANGE,
            buff=0.1,
        )  
        label15 = MathTex("x", r"\to", r"-\infty", color =ORANGE).next_to(arrow5, UP)


        arrow55 = Arrow(
            start=axes.c2p(-6, -6),
            end=axes.c2p(-6, -0.1),  
            color=ORANGE,
            buff=0.1,
        )  
        label155 = MathTex("y", r"\to", r"0^-", color =ORANGE).next_to(arrow55, LEFT)   



        # Graph of the function y = 1/(x+2) from x = 0 to x = 5
        graph1 = axes.plot(lambda x: 1/(x + 2), x_range=[0, 5], color=BLUE)
        graph1_label = MathTex(r"y = \frac{1}{x+2}", color=BLUE).next_to(graph1, UP * 0.5 + RIGHT).scale(0.7)   

        # Graph of the function y = 1/(x+2) from x = 0 to x = -1.9
        graph2 = axes.plot(lambda x: 1/(x + 2), x_range=[-1.9, 0], color=BLUE)

        # Graph of the function y = 1/(x+2) from x = -5 to x = -2.1
        graph3 = axes.plot(lambda x: 1/(x + 2), x_range=[-3, -2.1], color=BLUE) 

        #Graph of the function y = 1/(x+2) from x = -3 to x = -5
        graph4 = axes.plot(lambda x: 1/(x + 2), x_range=[-5, -3], color=BLUE)

    








    #####################################################################################################################################

        # Sequence of animations 

        self.play(Create(axes), run_time=2)
        self.wait(1)
        self.play(Create(y_axis_intercept), Write(y_axis_label), run_time=2)
        self.wait(2)

        # Draw asymptote y=0
        self.play(Create(asymptote), run_time=2)
        self.play(Write(asymptote_label), Write(text_asymptote), run_time=1)
        self.wait(2) 
        self.play(FadeOut(asymptote_label), FadeOut(text_asymptote))
        self.wait(1)  


        # Shake animation for asymptote y=0
        for _ in range(3):
            self.play(asymptote.animate.shift(UP * 0.05), run_time=0.05)
            self.play(asymptote.animate.shift(DOWN * 0.05), run_time=0.05)
        self.wait(2)

        # Draw asymptote x = -2
        self.play(Create(asymptote1), run_time=2)
        self.play(Write(asymptote1_label), Write(text_asymptote1), run_time=1)
        self.wait(2) 
        self.play(FadeOut(asymptote1_label), FadeOut(text_asymptote1))
        self.wait(1)


        # Shake animation for asymptote x = -2
        for _ in range(3):
            self.play(asymptote1.animate.shift(RIGHT * 0.05), run_time=0.05)
            self.play(asymptote1.animate.shift(LEFT * 0.05), run_time=0.05)
        self.wait(2)

        # Add arrow and label for x>0 region
        self.play(Create(arrow1), Write(label11), run_time=2)
        self.play(Create(arrow11), Write(label111), run_time=2)
        self.wait(3)
        
        
        

        # Draw graph of the function y = 1/(x+2) from x = 0 to x = 5
        self.play(Create(graph1), run_time=2)
        self.wait(2)
        self.play(Write(graph1_label), run_time=1)
        self.wait(2)
        self.play(FadeOut(arrow1), FadeOut(label11), FadeOut(arrow11), FadeOut(label111), run_time=2)


        # Add arrow and label for x tends to -2 from right side
        self.play(Create(arrow2), Write(label12), run_time=2)
        self.wait(3)
        self.play(Create(arrow22), Write(label122), run_time=2)
        self.wait(3)
        self.play(FadeOut(arrow2), FadeOut(label12), run_time=2)
        self.play(FadeOut(arrow22), FadeOut(label122), run_time=2)

        # Draw graph of the function y = 1/(x+2) from x = 0 to x = -1.9
        self.play(Create(graph2), run_time=2)
        self.wait(2)
        self.play(Write(graph1_label), run_time=1)
        self.wait(2)

        # Add point (-3, -1)
        self.play(Create(extra_point), Write(extra_point_label), run_time=2)
        self.wait(2)

        # Add arrow and label for x tends to -2 from left side and y tends to -infinity
        self.play(Create(arrow4), Write(label14), run_time=2)
        self.wait(2)
        self.play(Create(arrow44), Write(label144), run_time=2)
        self.wait(2)
        self.play(FadeOut(arrow4), FadeOut(label14), run_time=2)
        self.play(FadeOut(arrow44), FadeOut(label144), run_time=2)

        # Draw graph of the function y = 1/(x+2) from x = -3 to x = -2.1
        self.play(Create(graph3), run_time=2)
        self.wait(2)

        # Add arrow and label for x tends to -infinity  from left side and y tends to 0 from negative side
        self.play(Create(arrow5), Write(label15), run_time=2)
        self.wait(2)
        self.play(Create(arrow55), Write(label155), run_time=2)
        self.wait(2)
        self.play(FadeOut(arrow5), FadeOut(label15), run_time=2)
        self.play(FadeOut(arrow55), FadeOut(label155), run_time=2)

        # Draw graph of the function y = 1/(x+2) from x = -3 to x = -5
        self.play(Create(graph4), run_time=2)
        self.wait(2)
 
       
        





```

### Drawing SinX between 0 to 2PI

```python
%%manim -v WARNING -ql -i TrigGraph


class TrigGraph(Scene):

    def construct(self):

        # Create axes
        axes = Axes(
            x_range=[0, 2 * PI, PI / 2],
            y_range=[-1.5, 1.5, 0.5],
            x_length=10,
            y_length=6,
            axis_config={"color": WHITE},
            tips=False
        )

        axes.scale(0.8)
        axes.move_to(ORIGIN)

        # Add axis labels (custom ticks)
        x_labels = {
            #0: MathTex("0"),
            PI/2: MathTex(r"\frac{\pi}{2}"),
            PI: MathTex(r"\pi"),
            3*PI/2: MathTex(r"\frac{3\pi}{2}"),
            2*PI: MathTex(r"2\pi")
        }

        y_labels = {
            -1: MathTex("-1"),
             0: MathTex("0"),
            1: MathTex("1")
        }
        
        sine_label = MathTex(r"y = \sin(x)", color=BLUE).next_to(axes, UP * 3 + RIGHT * 3).scale(0.9)


        #  Important points on the graph of sine function
        important_points = [
            Dot(axes.c2p(0, 0), color=RED),
            Dot(axes.c2p(PI/2, 1), color=RED),
            Dot(axes.c2p(PI, 0), color=RED),
            Dot(axes.c2p(3*PI/2, -1), color=RED),
            Dot(axes.c2p(2*PI, 0), color=RED)
        ]  


        # plot the labels for important points
        important_point_labels = [
            MathTex("(0, 0)").next_to(important_points[0], DOWN + RIGHT * 0.1, ).scale(0.7).set_color(RED),
            MathTex(r"\left(\frac{\pi}{2}, 1\right)").next_to(important_points[1], UP + RIGHT * 0.1).scale(0.7).set_color(RED),
            MathTex(r"(\pi, 0)").next_to(important_points[2], DOWN + LEFT * 0.1).scale(0.7).set_color(RED),
            MathTex(r"\left(\frac{3\pi}{2}, -1\right)").next_to(important_points[3], DOWN + LEFT * 0.1).scale(0.7).set_color(RED),
            MathTex(r"(2\pi, 0)").next_to(important_points[4], DOWN + RIGHT * 0.1).scale(0.7).set_color(RED)
        ] 


        # Arrow for x increases from 0 to pi/2 then y increases from 0 to 1 with label
        arrow1 = Arrow(
            start=axes.c2p(0, 0),
            end=axes.c2p(PI/2, 0),
            color=GREEN,
            buff=0.1,
        )
        label_arrow1 = MathTex("x", r"\to", r"\frac{\pi}{2}", color=GREEN).move_to(axes.c2p(1, 0.3)).scale(0.7)

        arrow11 = Arrow(
            start=axes.c2p(0, 0),
            end=axes.c2p(0, 1),
            color=GREEN,
            buff=0.1,
        )
        label_arrow11 = MathTex("y", r"\to", r"1", color=GREEN).next_to(arrow11, LEFT).scale(0.8).rotate(PI/2)


        #Arrow for x increases from pi/2 to pi then y decreases from 1 to 0 with label
        arrow2 = Arrow(
            start=axes.c2p(PI/2, 0),
            end=axes.c2p(PI, 0),
            color=GREEN,
            buff=0.1,
        )  
        label_arrow2 = MathTex("x", r"\to", r"\pi", color=GREEN).next_to(arrow2, UP).scale(0.7)

        arrow22 = Arrow(
            start=axes.c2p(PI, 1),
            end=axes.c2p(PI, 0),
            color=GREEN,
            buff=0.1,
        )  
        label_arrow22 = MathTex("y", r"\to", r"0", color=GREEN).next_to(arrow22, RIGHT).scale(0.8).rotate(2*PI)



        # TEXT
        text_title = Text("In the similar manner", font_size=48).to_edge(UR)


        # creating Vgroup for all labels and points to manage them together if needed
        all_labels_and_points = VGroup()
        for label in important_point_labels:
            all_labels_and_points.add(label)
        for point in important_points:
            all_labels_and_points.add(point)
        
        #Creating for sine graph and their labels
        sine_graph_and_label = VGroup()
        sine_graph_and_label.add(sine_label)
        

    





        ################################################################################################################################

        # Sequence of animations

        # Apply labels
        axes_labels = axes.get_axis_labels(x_label="x", y_label="y")
        self.play(Create(axes))
        self.play(Write(axes_labels))

        # Add tick labels
        for x_val, label in x_labels.items():
            label.next_to(axes.c2p(x_val, 0), DOWN)
            self.play(Write(label),run_time=0.5)

        for y_val, label in y_labels.items():
            label.next_to(axes.c2p(0, y_val), LEFT)
            self.play(Write(label),run_time=0.5)

        self.wait()


        
        # Plot the point (0,0) on the graph and label it
        for point in important_points:
            self.play(Create(point), run_time=0.5)
            self.play(Write(important_point_labels[0]), run_time=0.5)
            self.wait(1)
            break        # i have added break here to stop after plotting first point (0,0) because the next code cell will plot all points again

        # Plot all those points where y value is 0  and label them
        i=0
        for point in important_points:

            if i!=0:
                if point.get_center()[1] == axes.c2p(0, 0)[1]:  # Check if y-coordinate is 0
                    self.play(Create(point), run_time=0.5)
                    self.play(Write(important_point_labels[i]), run_time=0.5)
                    self.wait(1)
            i = i+1
        
        #plot all the points where y value is maximum and minimum and label them
        i=0
        for point in important_points:
            
            if i!=0:
                if abs(point.get_center()[1]) == axes.c2p(0, 1)[1]:  # Check if y-coordinate is 1 or -1
                    self.play(Create(point), run_time=0.5)
                    self.play(Write(important_point_labels[i]), run_time=0.5)
                    self.wait(1)
            i = i+1

        # Since there are a lot of labels we fade out pi/2 and 3pi/2 labels to reduce clutter
        self.play(FadeOut(x_labels[PI/2]), FadeOut(x_labels[3*PI/2]), run_time=1)
        self.wait(1)  

        # Add arrow and label for x increases from 0 to pi/2 then y decreases from 1 to 0
        self.play(Create(arrow1), Write(label_arrow1), run_time=2)
        self.play(Create(arrow11), Write(label_arrow11), run_time=2)  
        self.wait(1)  

        # Draw the sine curve from 0 to pi/2
        sine_curve1 = axes.plot(lambda x: np.sin(x), x_range=[0, PI/2], color=BLUE)
        self.play(Create(sine_curve1), run_time=2)
        self.wait(2)

        # Fadeout the arrow and the label
        self.play(FadeOut(arrow1), FadeOut(label_arrow1), FadeOut(arrow11), FadeOut(label_arrow11), run_time=2)
        self.wait(1)

        # Add arrow and label for x increases from pi/2 to pi then y decreases from 1 to 0
        self.play(Create(arrow2), Write(label_arrow2), run_time=2)
        self.play(Create(arrow22), Write(label_arrow22), run_time=2)  
        self.wait(1)

        # Draw the sine curve from pi/2 to pi
        sine_curve2 = axes.plot(lambda x: np.sin(x), x_range=[PI/2, PI], color=BLUE)
        self.play(Create(sine_curve2), run_time=2)
        self.wait(2)

        # Fadeout the arrow and the label
        self.play(FadeOut(arrow2), FadeOut(label_arrow2), FadeOut(arrow22), FadeOut(label_arrow22), run_time=2)
        self.wait(2)


        # Add TEXT
        self.play(Write(text_title), run_time=2)
        self.wait(2)


        # draw the curve for rest of the interval from pi to 2pi
        sine_curve3 = axes.plot(lambda x: np.sin(x), x_range=[PI, 2*PI], color=BLUE)
        self.play(Create(sine_curve3), run_time=2)
        self.wait(2)

        # Fade out the text
        self.play(FadeOut(text_title), run_time=1)

        #Fade out label pi and 2pi to reduce clutter
        self.play(FadeOut(x_labels[PI]), FadeOut(x_labels[2*PI]), run_time=1)
        self.wait(2)

        # Final sine label
        self.play(Write(sine_label), run_time=2)
        self.wait(2)


```
