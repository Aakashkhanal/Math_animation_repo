### In this Manim  We learn about how to make exponential and logarithm curve.

```python
from manim import *
```

```python

%%manim -v WARNING -ql  -i Exponential_Curve_Sketching
class Exponential_Curve_Sketching(MovingCameraScene):
    def construct(self):

        # Divide the screen visually: left for text, right for graph
        left_text_area = Rectangle(width=6.5, height=7, color=GREY, fill_opacity=0).to_edge(LEFT)
        right_graph_area = Rectangle(width=6.5, height=7, color=YELLOW_B, fill_opacity=0).to_edge(RIGHT)


        # Uncomment next lines if you want to see the separation visually
        self.add(left_text_area, right_graph_area)

        # Axes for  graph 
        axes = Axes(
            x_range=[-5, 5, 1],
            y_range=[-5, 5, 1],
            x_length=10,
            y_length=10,
            axis_config={"color": WHITE},
            tips=False
        )
        
        axes.scale(0.5)  # ⬅️ scales down everything (axes + plotted graph)
        axes.to_edge(RIGHT, buff=1)  # ⬅️ moves it to the right half
############################################################################################################################################################################################

        # Graph of y = 2^x
        graph = axes.plot(lambda x: 2**x, color=BLUE, x_range=[-5, 5])  
        graph_label = axes.get_graph_label(graph, label="y = 2^x", x_val=2, direction=UP * 2 + 2*RIGHT)   
        self.play(Create(axes), Create(graph), Write(graph_label))
        self.wait(2)  


        # Plot key points when graph cuts y-axis
        y_axis_intercept = Dot(axes.c2p(0, 1), color=RED)
        random_point = Dot(axes.c2p(2, 4), color=RED)

        y_axis_label = MathTex("(0, 1)").next_to(y_axis_intercept, RIGHT)
        random_point_label = MathTex("(2, 4)").next_to(random_point, RIGHT)
        self.play(Create(y_axis_intercept), Write(y_axis_label))
        self.play(Create(random_point), Write(random_point_label))
        self.wait(2)


        # Asymptote line y=0
        asymptote = DashedLine(start=axes.c2p(-6, 0), end=axes.c2p(6, 0), color=YELLOW)
        asymptote_label = MathTex("y = 0 ").next_to(asymptote, DOWN).scale(0.7)
        self.play(Create(asymptote  ), Write(asymptote_label))
        self.wait(2)   

        # Shake animation

        for _ in range(3):  # repeat small shakes
            self.play(asymptote.animate.shift(UP * 0.05), run_time=0.05)
            self.play(asymptote.animate.shift(DOWN * 0.05), run_time=0.05)

        self.wait(2) 

        # Zoom in near the asymptote
        zoom_point = axes.c2p(-3, 0.5)
        self.play(self.camera.frame.animate.set(width=2).move_to(zoom_point), run_time=2)
        self.wait(2)

        # Optional: emphasize the gap between curve and asymptote
        arrow = Arrow(
            start=axes.c2p(-3, 0.1),
            end=axes.c2p(-3, 0.3),
            color=RED,
            buff=0.1,
        )
        self.play(GrowArrow(arrow), run_time=2)


        # Zoom out again to full view
        self.play(self.camera.frame.animate.set(width=14).move_to(ORIGIN), run_time=2)
        self.wait(1) 

        note = Text("Graph gets close\n but never touches!", color=YELLOW, font_size=24).next_to(arrow, UP)
        self.play( Write(note))
        self.wait(2)





        

```

```python
%%manim -v WARNING -ql  -i Exponential_Curve_Sketching1
class Exponential_Curve_Sketching1(MovingCameraScene):
    def construct(self):

        # Axes for graph 
        axes = Axes(
            x_range=[-5, 5, 1],
            y_range=[-5, 5, 1],
            x_length=10,
            y_length=10,
            axis_config={"color": WHITE},
            tips=False
        )
        
        axes.scale(0.8)  # Slightly smaller so it fits nicely
        axes.move_to(ORIGIN)
        ############################################################################################################################################################################################

        # Graph of y = 2^x
        graph = axes.plot(lambda x: 2**x, color=BLUE, x_range=[-5, 0])  
        graph_label = axes.get_graph_label(
            graph, label="y = 2^x", x_val=2, direction=UP * 2 + 10 * RIGHT
        )

        graph1 = axes.plot(lambda x: 2**x, color=BLUE, x_range=[0, 5])  
        
        
   

        # Plot key points when graph cuts y-axis
        y_axis_intercept = Dot(axes.c2p(0, 1), color=RED)
        random_point = Dot(axes.c2p(2, 4), color=RED)

        y_axis_label = MathTex("(0, 1)").next_to(y_axis_intercept, RIGHT)
        random_point_label = MathTex("(2, 4)").next_to(random_point, RIGHT)
       
       
        self.wait(2)

        # Asymptote line y=0
        asymptote = DashedLine(start=axes.c2p(-6, 0), end=axes.c2p(6, 0), color=YELLOW)
        asymptote_label = MathTex("y = 0").next_to(asymptote, DOWN).scale(0.7)




        # Zoom in near the asymptote
        zoom_point = axes.c2p(-3, 0.5)
  

        # Emphasize the gap between curve and asymptote (with curly brace)
        brace = BraceBetweenPoints(
            axes.c2p(-3, 0.01), axes.c2p(-3, 0.09), direction=LEFT, color=RED
        )
   
        

      

        # Add note explaining the gap
        note = Text(
            "Graph gets close\nbut never touches!",
            color=RED_B,
            font_size=24
        ).next_to(brace, LEFT+ 2*UP)

        # Add arrow pointing to x>0 region and y>0 region
        
        arrow1 = Arrow(
            start=axes.c2p(1, -1),
            end=axes.c2p(5, -1),
            color=GREEN,
            buff=0.1,
           
        )  
        label11 = MathTex("x > 0", color=ORANGE).next_to(arrow1, RIGHT) 

        arrow2 = Arrow(
            start=axes.c2p(-1, 2),
            end=axes.c2p(-1, 5),
            color=GREEN,
            buff=0.1,
            
        )
        label12 = MathTex("y > 0", color=ORANGE).next_to(arrow2, LEFT)

         # Add arrow pointing to x<0 region 
        arrow3 = Arrow(
            start=axes.c2p(-1, -1),
            end=axes.c2p(-5, -1),
            color=GREEN,
            buff=0.1,
            
        )
        label13 = MathTex("x < 0", color=ORANGE).next_to(arrow3, LEFT)

         

        # Shade the region where y < 0
        y_neg_region = Polygon(
        axes.c2p(-7, 0),  # left x-axis intercept
        axes.c2p(7, 0),   # right x-axis intercept
        axes.c2p(7, -7),  # bottom right
        axes.c2p(-7, -7), # bottom left
        color=GRAY,
        fill_opacity=0.2,
        stroke_width=0
    )

        # text for the shaded region

        text_y_neg = Text(
            "y < 0 region\n(not part of graph)",
            color=WHITE,
            font_size=30
        ).move_to(axes.c2p(-3, -3)) 
        




    #####################################################################################################################################

        # Sequence of animations 
        self.play(Create(axes))
        self.wait(1)
        self.play(Create(y_axis_intercept), Write(y_axis_label))
        self.wait(2)

        self.play( Create(graph), Write(graph_label))
        self.wait(2)

        self.play(Create(asymptote), Write(asymptote_label))
        self.wait(2)

        # Shake animation for asymptote
        for _ in range(3):
            self.play(asymptote.animate.shift(UP * 0.05), run_time=0.05)
            self.play(asymptote.animate.shift(DOWN * 0.05), run_time=0.05)
        self.wait(2)

        # Zoom in near the asymptote
        self.play(self.camera.frame.animate.set(width=2).move_to(zoom_point), run_time=2)
        self.wait(2)

        #Emphasize the gap between curve and asymptote 
        self.play(Create(brace), run_time=1.5)
        self.wait(1)

        # Zoom out again to full view
        self.play(self.camera.frame.animate.set(width=14).move_to(ORIGIN), run_time=2)
        self.wait(1)


        # Gap note 
        self.play(Write(note))
        self.wait(2)

        #Plot the point (2,4) and rest of the graph from x=0 to x=5
        self.play(Create(random_point), Write(random_point_label))
        self.play(Create(graph1), run_time=2)
        self.wait(2)

        # Show arrows pointing to different regions
        self.play(GrowArrow(arrow1), GrowArrow(arrow2), GrowArrow(arrow3), run_time=2)
        self.play(Write(label11), Write(label12), Write(label13), run_time=2)
        self.wait(3) 

        self.play(FadeOut(label11), FadeOut(label12), FadeOut(label13), FadeOut(arrow1), FadeOut(arrow2), FadeOut(arrow3), run_time=1)


        # Shade the region where y < 0
        self.play(Create(y_neg_region), run_time=2)
        self.wait(2)
        self.play(Write(text_y_neg), run_time=2)
        self.wait(3)

```
